## 1. How much movies were added each year and plot the growth,do the same for tv-shows 

```python
netflix.columns
```

```python
movies_growth = netflix[['type', 'date_added']]
movies_growth.head()
```

```python
movies_growth['date_added'] = movies_growth['date_added'].astype(str).str.strip()
movies_growth['date_added'] = pd.to_datetime(movies_growth['date_added'], errors="coerce")
movies_growth['Month'] = movies_growth['date_added'].dt.strftime('%B')
movies_growth['Day'] = movies_growth['date_added'].dt.strftime("%d")
movies_growth[ 'Year'] = movies_growth["date_added"].dt.year

print(movies_growth)
```

```python
movies_growth.head()
```

```python
movies_df = movies_growth[movies_growth['type'] == 'Movie'].copy()
tv_shows_df = movies_growth[movies_growth['type'] == 'TV Show'].copy()
```

```python
movies_df.head()
```

```python
movies_df.groupby(['Year'])
```

```python
movies_by_year = movies_df['Year'].value_counts().sort_index().astype(int)
movies_by_year.head(50)
```

```python
import mplcursors

#The bar chart
plt.figure(figsize=(10,6))
bars = movies_by_year.plot(kind='bar', color='skyblue')

# Add titles and labels
plt.title('Number of Movies per Year')
plt.xlabel('Year')
plt.ylabel('Number of Movies')
plt.xticks(rotation=45)

#Hover mouse over for the numbers
#cursor = mplcursors.cursor(bars, hover=True)

# cursor.connect(
#     "add", lambda sel: sel.annotation.set_text(f'{sel.artist.get_height()} Movies'))

plt.show()
plt.tight_layout()
```

```python
tv_shows_by_year = tv_shows_df['Year'].value_counts().sort_index()
tv_shows_by_year.head()
```

```python
plt.figure(figsize=(10,6))
tv_shows_by_year.plot(kind='bar', color='red', 
                     title='Number of Tv shows per year',
                     xlabel='Year',
                     ylabel='Number of Tv shows')

plt.show()
```

## 2. Who was the director of the most series/tv shows, create a top 10

```python
netflix.head()
director_type = netflix[['type', 'director']]
director_type.dropna()
```

```python
# Devide the tv shows and the movies in a seperate df 
director_movies = director_type[director_type['type']== 'Movie'].copy()
director_tvshow = director_type[director_type['type'] == 'TV Show'].copy()

# Count the number of movies/tv shows per director
director_shows_by_year = director_tvshow["director"].value_counts().nlargest(10).sort_index()
director_movies_by_year = director_movies["director"].value_counts().nlargest(10).sort_index()
```

```python
director_movies_by_year.plot(kind="bar", figsize=(10,6), color='Skyblue',
                           title='Top 10 directors by number of movies',
                           xlabel="Director",
                           ylabel='Number of movies')

plt.show()
```

```python
director_shows_by_year.nlargest(10).plot(kind="bar", figsize=(10,6), color="darkblue",
                           title='Top 10 directors number of shows ',
                           xlabel='Directors',
                           ylabel='Number of shows')

plt.show()
```

## 3. Create a top 10 of the countries with the most tv-shows and movies at the top.

```python
netflix.head()
```

```python
country_top = netflix[['type', 'country','release_year']]
country_top.dropna().head()
```

```python
# Separate the movies from the tv shows 
country_top_tvshows = country_top[country_top['type'] == 'TV Show'].copy()
country_top_movies = country_top[country_top['type'] == 'Movie'].copy()

# Fill in the NAN values with No country
country_top_tvshows['country']= country_top_tvshows['country'].fillna('No country')
country_top_movies['country']= country_top_movies['country'].fillna('No country')
country_top_movies.head(10)
```

```python
country_top10 = country_top_movies['country'].value_counts().nlargest(10).sort_index()

country_top10.plot(kind='bar', figsize=(10,6), color='Blue',
                       title='Top 10 countrys with the most movies',
                       xlabel='Country',
                       ylabel='Total')

plt.show()
```

```python
country_shows_top10 = country_top_tvshows['country'].value_counts().nlargest(10).sort_index()

country_shows_top10.plot(kind='bar', figsize=(10,6), color='green',
                       title='Top 10 countrys with the most tv shows',
                       xlabel='Country',
                       ylabel='Total')

plt.show()
```

## 4. Create a top 10 of longest duration of movies. 
Do the same for the tv-shows but then rank them by seasons

```python
netflix.head()
```

```python
duration_movies = netflix[netflix['type']== 'Movie']

#drop the NAN lines 
duration_movies = duration_movies.dropna(subset=['duration']) #remove rows where duration = NaN

# Remove the min from duration 
duration_movies['duration'] = duration_movies['duration'].str.replace('min', '').str.strip()
duration_movies['duration'] = duration_movies['duration'].astype(int)

duration_movies.head()
```

```python
duration_movies = duration_movies[['title', 'duration']]
duration_movies.head()
```

```python
topduration_movies = duration_movies.sort_values(['duration'], ascending=False)

topduration_movies.nlargest(10, 'duration').plot(kind='bar',x='title', y='duration', figsize=(10,6), color='Blue',
                       title='Longest movies',
                       xlabel='Title',
                       ylabel='Duration')

plt.show()
```

```python
duration_tvshow = netflix[netflix['type']== 'TV Show']

#drop the NAN lines 
duration_movies = duration_movies.dropna(subset=['duration']) #remove rows where duration = NaN

# Remove the min from duration 
duration_tvshow['duration'] = duration_tvshow['duration'].str.replace('Seasons', '').str.strip()
duration_tvshow['duration'] = duration_tvshow['duration'].str.replace('Season', '').str.strip()
duration_tvshow['duration'] = duration_tvshow['duration'].astype(int)

duration_tvshow.head()
```

```python
topduration_tvshow = duration_tvshow.sort_values(['duration'], ascending=False)

topduration_tvshow.nlargest(10, 'duration').plot(kind='bar',x='title', y='duration', figsize=(10,6), color='Blue',
                       title='Longest TV Shows',
                       xlabel='TV Show',
                       ylabel='Duration')

plt.show()
```
